```python
import random
from collections import defaultdict

class MarkovTextGenerator:
    def __init__(self, order=2):
        self.order = order
        self.model = defaultdict(list)

    def train(self, text):
        """Train the model on input text"""
        words = text.split()
        for i in range(len(words) - self.order):
            key = tuple(words[i:i + self.order])
            self.model[key].append(words[i + self.order])

    def generate(self, length=100, seed=None):
        """Generate new text"""
        if not self.model:
            return "Model not trained yet!"

        # Start with a random seed
        if seed is None:
            seed = random.choice(list(self.model.keys()))
        else:
            seed = tuple(seed.split()[:self.order])

        current = list(seed)
        result = current[:]

        for _ in range(length):
            key = tuple(current)
            if key not in self.model:
                break
            next_word = random.choice(self.model[key])
            result.append(next_word)
            current = current[1:] + [next_word]

        return ' '.join(result)

# Example usage
if __name__ == "__main__":
    # Sample training text (you can replace with a large corpus)
    sample_text = """
    The quick brown fox jumps over the lazy dog.
    Artificial intelligence is transforming the world.
    Python is a great programming language for text generation.
    Machine learning models can generate human-like text.
    Deep learning requires lots of data and computing power.
    """

    generator = MarkovTextGenerator(order=2)
    generator.train(sample_text)

    print("Generated Text:")
    print(generator.generate(length=50))
```
